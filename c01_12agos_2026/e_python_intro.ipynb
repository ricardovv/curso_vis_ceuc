2- Decisiones, if, else, elif
</div>

* Es la manera en que el comoutador toma desiciones, preguntando si un valor es tal o cual.

```python
## Probar cambiando el valor de a
a = 5

if a == 5:
    print("a vale 5")
else:
    print("a NO vale 5")
```

3- Ciclos
</div>

* Permite repetir instrucciones según un límite.

```python
## Para cada elemento i, entre 0 y 6, se hace lo siguiente...

# for(int i=0; i<3; i++ ){
#     hacer algo...
# }


for i in range( 2, 10 ):
    print( "i Vale: ", i )
```

```python
## Se pueden hacer operaciones a cada elemento
val = 3
for i in range( 10 ):
    print( "sumar val más", i, "es:", val + i )
```

4- Listas o Arrays
</div>

* Es una estructura de datos que sirve para tener una lista de muchos valores guardados con un mismo nombre.

```python
## En este caso son textos que van entre comillas.

nombres = ['Ana','Beatriz','Claudia','Diana', 'Marcianeke']
edades  = [34, 23, 3, 12, 5]

print( nombres[2] , edades[2] ) ## cambiar valor entre los paréntesis cuadrados para acceder a otro nombre de la lista
```

<div style="background:deepskyblue; color:aliceblue; font-size:28px; padding:30px; text-align:left; font-weight:bold; ">
5- Recorrer listas
</div>

* Se puede recorrer una lista con un ciclo.

```python
## Toma cada elemento n de la lista nombres

for n in range( len(nombres) ): ## len(nombres) indica la cantidad de elementos de la lista nombres
    print( nombres[ n ] )
```

```python
## Podemos tener listas de números también!

edades = ['cla',23,18,54,29,43,41,33]

l = len(edades)

for x in range( l ): ## len(nombres) indica la cantidad de elementos de la lista nombres
    print( edades[ x ] )
```
